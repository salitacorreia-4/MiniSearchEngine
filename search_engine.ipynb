```python
def process_query(query):
    tokens = tokenize(query)
    tokens = remove_stop_words(tokens)
    tokens = stem_tokens(tokens)

    return tokens
```

```python
query = "Python machine learning"

print(process_query(query))
```

```python
def search(query):
    query_tokens = process_query(query)

    matching_documents = set()

    for token in query_tokens:
        if token in inverted_index:
            matching_documents.update(inverted_index[token])

    return matching_documents
```

```python
def display_results(query):
    results = search(query)

    print("Search:", query)
    print("Number of results:", len(results))
    print()

    for i, document in enumerate(results, start=1):
        print(i, ".", document)
```

```python
def calculate_tf(word, tokens):
    count = tokens.count(word)

    if len(tokens) == 0:
        return 0

    return count / len(tokens)
```

```python
tokens = stemmed_documents["python.txt"]

print(calculate_tf("python", tokens))
```

```python
import math
```

```python
def calculate_idf(word):
    total_documents = len(stemmed_documents)

    document_count = 0

    for tokens in stemmed_documents.values():
        if word in tokens:
            document_count += 1

    if document_count == 0:
        return 0

    return math.log(total_documents / document_count)
```

```python
print(calculate_idf("python"))
```

```python
print(calculate_idf("java"))
```

```python
def calculate_tfidf(word, tokens):
    tf = calculate_tf(word, tokens)
    idf = calculate_idf(word)

    return tf * idf
```

```python
tokens = stemmed_documents["python.txt"]

print(calculate_tfidf("python", tokens))
```

```python
def document_score(query, document_name):
    query_tokens = process_query(query)
    document_tokens = stemmed_documents[document_name]

    score = 0

    for word in query_tokens:
        score += calculate_tfidf(word, document_tokens)

    return score
```

```python
print(document_score("python machine learning", "python.txt"))
```

```python
def ranked_search(query):
    results = search(query)

    scored_results = []

    for document in results:
        score = document_score(query, document)
        scored_results.append((document, score))

    scored_results.sort(key=lambda x: x[1], reverse=True)

    return scored_results
```

```python
results = ranked_search("python machine learning")

print(results)
```

```python
def display_ranked_results(query):

    results = ranked_search(query)

    print("=" * 40)
    print("MINI SEARCH ENGINE")
    print("=" * 40)

    print("Search:", query)
    print("Results:", len(results))
    print()

    for i, (document, score) in enumerate(results, start=1):

        print(f"{i}. {document}")
        print(f"   Score: {score:.4f}")
        print()
```

```python
vocabulary = set()

for tokens in stemmed_documents.values():
    vocabulary.update(tokens)

vocabulary = sorted(vocabulary)

print(vocabulary)
```

```python
def tfidf_vector(document_name):
    
    tokens = stemmed_documents[document_name]
    
    vector = []

    for word in vocabulary:
        score = calculate_tfidf(word, tokens)
        vector.append(score)

    return vector
```

```python
vector = tfidf_vector("python.txt")

print(vector)
```

```python
def query_vector(query):
    
    tokens = process_query(query)
    
    vector = []

    for word in vocabulary:
        score = calculate_tf(word, tokens) * calculate_idf(word)
        vector.append(score)

    return vector
```

```python
vector = query_vector("python machine learning")

print(vector)
```

```python
def cosine_similarity(vector1, vector2):
    
    dot_product = 0
    magnitude1 = 0
    magnitude2 = 0

    for i in range(len(vector1)):
        dot_product += vector1[i] * vector2[i]
        magnitude1 += vector1[i] ** 2
        magnitude2 += vector2[i] ** 2

    magnitude1 = math.sqrt(magnitude1)
    magnitude2 = math.sqrt(magnitude2)

    if magnitude1 == 0 or magnitude2 == 0:
        return 0

    return dot_product / (magnitude1 * magnitude2)
```

```python
query = "python machine learning"

q_vector = query_vector(query)
d_vector = tfidf_vector("python.txt")

similarity = cosine_similarity(q_vector, d_vector)

print(similarity)
```

```python
d_vector = tfidf_vector("java.txt")

similarity = cosine_similarity(q_vector, d_vector)

print(similarity)
```

```python
def cosine_ranked_search(query):
    
    results = search(query)

    query_vec = query_vector(query)

    scored_results = []

    for document in results:
        
        document_vec = tfidf_vector(document)
        
        score = cosine_similarity(query_vec, document_vec)
        
        scored_results.append((document, score))

    scored_results.sort(key=lambda x: x[1], reverse=True)

    return scored_results
```

```python
results = cosine_ranked_search("python machine learning")

print(results)
```

```python
def final_search(query):

    results = cosine_ranked_search(query)

    print("=" * 50)
    print("             MINI SEARCH ENGINE")
    print("=" * 50)

    print("Query:", query)
    print()

    if len(results) == 0:
        print("No documents found.")
        return

    for i, (document, score) in enumerate(results, start=1):
        
        print(f"{i}. {document}")
        print(f"   Relevance Score: {score:.4f}")
        print()
```
